```python
import matplotlib.pyplot as plt
data=pd.read_csv("StudentsPerformance.csv")
data
```

```python
gender_count = data['gender'].value_counts()

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct='%1.1f%%'
)

plt.title('Percentage of Male and Female Students')
plt.show()
```

```python
group_count = data['race/ethnicity'].value_counts()
group_count = group_count.loc[['group A', 'group B', 'group C']]
plt.pie(group_count,labels=group_count.index,autopct='%1.1f%%',startangle=90)
plt.title('Percentage of Students in Groups A, B and C')
plt.show()
```

```python
d=pd.read_csv("planets.csv")
d
```

```python
plt.boxplot(d["orbital_period"].dropna())
plt.title("Box Plot of Orbital Period")
plt.ylabel("Orbital Period")
plt.show()
```

```python

columns = ["orbital_period", "mass", "distance"]

for col in columns:
    d[col] = pd.to_numeric(d[col], errors="coerce")
d[columns].plot(kind="box",subplots=True,layout=(1, 3),figsize=(12, 5),sharey=False)
plt.suptitle("Exoplanet Data - Box Plots")
plt.tight_layout()
plt.show()
```
