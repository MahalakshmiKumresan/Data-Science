```python
bill = ds["total_bill"]

print("Mean:", bill.mean())
print("Median:", bill.median())
print("Mode:", bill.mode().tolist())
print("Variance:", bill.var())
print("Standard deviation:", bill.std())
print("Minimum:", bill.min())
print("Maximum:", bill.max())
print("Q1:", bill.quantile(0.25))
print("Q3:", bill.quantile(0.75))
print("IQR:", bill.quantile(0.75) - bill.quantile(0.25))
```

```python
kurtosis = bill.kurt()

print("Excess kurtosis:", kurtosis)

if kurtosis > 0:
    print("Positive excess kurtosis.")
elif kurtosis < 0:
    print("Negative excess kurtosis.")
else:
    print("Zero excess kurtosis.")
```

```python
from scipy import stats
bill = ds["total_bill"].dropna()
stats.probplot(bill, dist="norm", plot=plt)
plt.title("Q-Q Plot of Restaurant Bills")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.show()
```

```python
d=pd.read_csv("Iris.csv")
stats.probplot(d["SepalLengthCm"],dist="norm",plot=plt)
plt.title("Normal  distribution QQ Plot - Iris Sepal Length")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(ds["total_bill"].dropna(), bins=15, edgecolor="black")
plt.title("Distribution of Restaurant Bills")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
np.random.seed(42)
data = np.random.exponential(scale=2, size=500)
stats.probplot(data, dist="norm", plot=plt)
plt.title("Q-Q Plot - Exponential Distribution")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.grid(True)
plt.show()
```
