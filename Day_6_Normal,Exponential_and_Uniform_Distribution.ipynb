```python
from scipy.stats import norm

# Given values
mu = 98.6
sigma = 0.5
X = 99.5

# Probability of temperature higher than 99.5
prob_greater = 1 - norm.cdf(X, mu, sigma)

print("Probability of body temperature higher than 99.5°F:",
      round(prob_greater * 100, 2), "%")
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Normal Distribution parameters
mu = 98.6
sigma = 0.5

# Generate random sample dataset
X = np.random.normal(mu, sigma, 10000)

# Histogram
plt.hist(X, bins=30, density=True, alpha=0.6)

# Plot normal distribution curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 1000)
y = norm.pdf(x, mu, sigma)

plt.plot(x, y, linewidth=2)

plt.title("Normal Distribution - Human Body Temperature")
plt.xlabel("Body Temperature (°F)")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

```python

from scipy.stats import expon

# Given values
mean = 15
scale = mean
X = 30

# Probability of waiting more than 30 minutes
prob_greater = 1 - expon.cdf(X, scale=scale)

print("Probability of waiting more than 30 minutes:",
      round(prob_greater * 100, 2), "%")
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

# Exponential Distribution parameters
mean = 15
scale = mean

# Generate random sample dataset
X = np.random.exponential(scale, 10000)

# Histogram
plt.hist(X, bins=30, density=True, alpha=0.6)

# Plot exponential distribution curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 1000)
y = expon.pdf(x, scale=scale)

plt.plot(x, y, linewidth=2)

plt.title("Exponential Distribution - Hospital Patient Arrival")
plt.xlabel("Time Between Patient Arrivals (Minutes)")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

```python

from scipy.stats import uniform

# Given values
low = 10
high = 30
X = 25

# Uniform Distribution parameters
loc = low
scale = high - low

# Probability of processing time greater than 25 minutes
prob_greater = 1 - uniform.cdf(X, loc=loc, scale=scale)

print("Probability of processing time exceeding 25 minutes:",
      round(prob_greater * 100, 2), "%")
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

# Uniform Distribution parameters
low = 10
high = 30

# Generate random sample dataset
X = np.random.uniform(low, high, 10000)

# Histogram
plt.hist(X, bins=30, density=True, alpha=0.6)

# Plot uniform distribution curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 1000)
y = uniform.pdf(x, loc=low, scale=high-low)

plt.plot(x, y, linewidth=2)

plt.title("Uniform Distribution - Medical Equipment Processing Time")
plt.xlabel("Processing Time (Minutes)")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```
