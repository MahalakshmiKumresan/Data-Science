```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Train_Data.csv")

data.head()
```

```python
x = data[['age']]
y = data['charges']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train, y_train)
```

```python
age = int(input("Enter your age: "))

experience = pd.DataFrame({'age': [age]})

prediction = model.predict(experience)

print("Predicted Insurance Charges:", prediction[0])
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score
acc = r2_score(y_test, y_pred)
print("R2 Score:", acc * 100)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_test, y_test, label="Actual Values")

plt.plot(x_test, y_pred, color='red', label="Predicted Line")

plt.xlabel("Age")
plt.ylabel("Insurance Charges")

plt.title("Age vs Insurance Charges")

plt.legend()
plt.show()
```

```python
data = pd.read_csv("student_scores.csv")

data.head()
```

```python
x = data[['Hours']]
y = data['Score']
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
hours = float(input("Enter your study hours: "))

study_hours = pd.DataFrame({'Hours': [hours]})

print("Predicted Score:", model.predict(study_hours))
```

```python
y_pred = model.predict(x_test)
```

```python
acc = r2_score(y_test, y_pred)
print("R2 Score:", acc * 100)
```

```python
plt.scatter(x_test, y_test, label="Actual Values")

plt.plot(
    x_test.sort_values(by='Hours'),
    model.predict(x_test.sort_values(by='Hours')),
    color='red',
    label="Predicted Line"
)

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.title("Study Hours vs Exam Scores")

plt.legend()
plt.show()
```
